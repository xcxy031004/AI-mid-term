# Keras深度學習初體驗1
利用keras套件，產生一個32個神經元的深度學習模型進行訓練


```
from tensorflow import keras # 匯入 Keras 深度學習函式庫
from keras.models import Sequential # 匯入順序式神經網路模型
from keras.layers import Dense # 匯入全連接層(Dense)

import numpy as np
data = np.random.random((1000, 10)) #隨機產生1000筆資料，每筆資料包含十個float數字
labels = np.random.randint(2, size=(1000, 1)) #隨機產生1000筆資料，每筆資料上限為2(意思就是0 或 1)
print(np.shape(data))
print(data[0])
print(np.shape(labels))
print(labels[0])

model = Sequential() # 建立順序式神經網路模型
model.add(Dense(32, activation='relu', input_dim=10)) # 隱藏層，共32個神經元 # 輸入資料有10個特徵 # 使用ReLU激活函數
model.add(Dense(1, activation='sigmoid')) # 輸出層，輸出0或1的分類結果
model.compile(optimizer='rmsprop', loss='binary_crossentropy', metrics=['accuracy']) # 設定模型訓練方式 # binary_crossentropy適用於二元分類問題 # accuracy作為評估指標

model.fit(data, labels, epochs=10, batch_size=32)
# 使用訓練資料訓練模型
# 共訓練10次
# 每次使用32筆資料更新權重
score=model.evaluate(data, labels) # 評估模型效能
print(score[1]) # 顯示模型準確率
```
